```python
import plotly.graph_objects as go


# Generic cross plot function
def create_cross_plot(
    x_data,
    y_data,
    x_label="X Axis",
    y_label="Y Axis",
    title="Cross Plot",
    width=None,
    height=None,
    lock_zoom_x=False,
    lock_zoom_y=False,
):
    """
    Create a scatter plot (cross plot) using Plotly.

    Parameters:
    - x_data: List or array of x values
    - y_data: List or array of y values
    - x_label: Label for x-axis
    - y_label: Label for y-axis
    - title: Title of the plot
    - width: Width of the figure in pixels (optional)
    - height: Height of the figure in pixels (optional)
    - lock_zoom_x: If True, lock the zoom on x-axis by setting fixed range
    - lock_zoom_y: If True, lock the zoom on y-axis by setting fixed range
    """
    fig = go.Figure()

    fig.add_trace(
        go.Scatter(
            x=x_data,
            y=y_data,
            mode="markers",
            marker={"size": 8, "color": "blue", "opacity": 0.6},
            name="Data Points",
        )
    )

    layout_kwargs = {
        "title": title,
        "template": "plotly_white",
        "width": width,
        "height": height,
    }

    xaxis_kwargs = {"title": x_label}
    if lock_zoom_x:
        xaxis_kwargs["fixedrange"] = True
        xaxis_kwargs["range"] = [min(x_data), max(x_data)]
    layout_kwargs["xaxis"] = xaxis_kwargs

    yaxis_kwargs = {"title": y_label}
    if lock_zoom_y:
        yaxis_kwargs["fixedrange"] = True
        yaxis_kwargs["range"] = [min(y_data), max(y_data)]
    layout_kwargs["yaxis"] = yaxis_kwargs

    fig.update_layout(**layout_kwargs)

    return fig
```

```python
# Example usage with dummy data
import numpy as np

x_example = np.random.randn(100)
y_example = np.random.randn(100)

fig = create_cross_plot(
    x_example,
    y_example,
    x_label="Generic X",
    y_label="Generic Y",
    title="Sample Cross Plot",
    width=600,
    height=600,
    lock_zoom_x=False,
    lock_zoom_y=True,
)
fig.show()
```
